# Exercise 6 — Max and the Fuel Crops


## Task

Max has access to **100 acres** of farmland and wants to grow corn and sunflowers. His investment is limited to **720 kEUR**. He also wants to cultivate sunflowers, but on at most **40 acres**.

| Parameter | Unit | Corn | Sunflowers |
|---|---|---:|---:|
| Labour and material cost | kEUR/acre | 6 | 9 |
| Profit | kEUR/acre | 1 | 2 |

Tasks:

1. Formulate the problem algebraically.
2. Translate it into JuMP and solve it.
3. Interpret the solution.


## Reference implementation


```julia
using JuMP, HiGHS

# data input
crops = [:corn, :sunflower]
cost = Dict(crops .=> [6, 9])
profit = Dict(crops .=> [1, 2])

# model
m = Model(HiGHS.Optimizer)
@variable(m, x[crops] >= 0) # non-negativity
@expression(m, total_acres, sum(x[c] for c in crops))
@constraint(m, max_acres, total_acres <= 100) # max acres

@expression(m, total_investment, sum(x[c]*cost[c] for c in crops))
@constraint(m, max_invest, total_investment <= 720) # maximum investment
@constraint(m, sunflowers, x[:sunflower] <= 40) # sunflower constraint
@expression(m, total_profit, sum(x[c]*profit[c] for c in crops))
@objective(m, Max, total_profit)
optimize!(m)
objective_value(m)
value.(x)
dual.(max_acres)
dual.(max_invest)
dual.(sunflowers)
```
